```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
```

```python
word_vectors = {
    'king':   np.array([0.82, 0.21, 0.76, 0.34]),
    'queen':  np.array([0.79, 0.25, 0.72, 0.39]),
    'prince': np.array([0.74, 0.18, 0.69, 0.30]),
    'apple':  np.array([0.12, 0.88, 0.16, 0.11]),
    'orange': np.array([0.16, 0.81, 0.19, 0.14]),
    'car':    np.array([0.43, 0.35, 0.22, 0.81])
}

for word, vector in word_vectors.items():
    print(f'{word:8} -> {vector}')
```

```python
dimension = len(word_vectors['king'])
print('Embedding dimension:', dimension)
```

```python
def word_similarity(word_a, word_b):
    vector_a = word_vectors[word_a].reshape(1, -1)
    vector_b = word_vectors[word_b].reshape(1, -1)
    return float(cosine_similarity(vector_a, vector_b)[0][0])

score = word_similarity('king', 'queen')
print(f'Cosine similarity between king and queen: {score:.4f}')
```

```python
pairs = [
    ('king', 'queen'),
    ('king', 'prince'),
    ('apple', 'orange'),
    ('king', 'apple'),
    ('apple', 'car')
]

for a, b in pairs:
    print(f'{a:8} vs {b:8} -> {word_similarity(a, b):.4f}')
```

```python
target = 'apple'

similar_words = []
for word in word_vectors:
    if word != target:
        similar_words.append((word, word_similarity(target, word)))

similar_words.sort(key=lambda x: x[1], reverse=True)

print(f'Most similar words to {target}:')
for word, score in similar_words:
    print(f'{word:8} -> {score:.4f}')
```

```python
for word, vector in word_vectors.items():
    magnitude = np.linalg.norm(vector)
    print(f'{word:8} -> magnitude = {magnitude:.4f}')
```

```python
relationship = word_vectors['king'] - word_vectors['prince']
print('king - prince =', np.round(relationship, 4))

candidate_scores = {}
for word, vector in word_vectors.items():
    if word not in {'king', 'prince'}:
        candidate_scores[word] = cosine_similarity(
            relationship.reshape(1, -1),
            vector.reshape(1, -1)
        )[0][0]

print('\nSimilarity of the relationship vector to remaining words:')
for word, score in sorted(candidate_scores.items(), key=lambda x: x[1], reverse=True):
    print(f'{word:8} -> {score:.4f}')
```

```python
words = list(word_vectors.keys())
matrix = np.array([word_vectors[word] for word in words])
similarity_matrix = cosine_similarity(matrix)

print('Words:', words)
print('\nCosine similarity matrix:')
print(np.round(similarity_matrix, 3))
```
