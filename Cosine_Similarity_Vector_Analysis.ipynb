```python
vector_X = [3, 1, 4]
vector_Y = [2, 5, 1]

print("Vector X:", vector_X)
print("Vector Y:", vector_Y)
```

```python
def calculate_dot_product(X, Y):
    result = 0
    for x, y in zip(X, Y):
        result += x * y
    return result

dot_value = calculate_dot_product(vector_X, vector_Y)
print("Dot Product (X · Y):", dot_value)
```

```python
import math

def calculate_magnitude(vector):
    squared_sum = 0
    for value in vector:
        squared_sum += value ** 2
    return math.sqrt(squared_sum)

magnitude_X = calculate_magnitude(vector_X)
magnitude_Y = calculate_magnitude(vector_Y)

print("Magnitude of X:", round(magnitude_X, 4))
print("Magnitude of Y:", round(magnitude_Y, 4))
```

```python
def calculate_cosine_similarity(X, Y):
    numerator = calculate_dot_product(X, Y)
    denominator = calculate_magnitude(X) * calculate_magnitude(Y)
    return numerator / denominator

manual_score = calculate_cosine_similarity(vector_X, vector_Y)
print("Cosine Similarity (Manual):", round(manual_score, 5))
```

```python
import numpy as np

X = np.array(vector_X)
Y = np.array(vector_Y)

numpy_score = np.dot(X, Y) / (np.linalg.norm(X) * np.linalg.norm(Y))
print("Cosine Similarity (NumPy):", round(numpy_score, 5))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

X_2d = X.reshape(1, -1)
Y_2d = Y.reshape(1, -1)

sklearn_score = cosine_similarity(X_2d, Y_2d)[0, 0]
print("Cosine Similarity (scikit-learn):", round(sklearn_score, 5))
```

```python
print("Manual method      :", round(manual_score, 6))
print("NumPy              :", round(numpy_score, 6))
print("scikit-learn       :", round(sklearn_score, 6))

assert np.isclose(manual_score, numpy_score)
assert np.isclose(numpy_score, sklearn_score)
print("\nAll methods produce the same cosine similarity.")
```

```python
test_cases = {
    "Same direction": ([2, 4], [1, 2]),
    "Opposite direction": ([2, 4], [-1, -2]),
    "Orthogonal vectors": ([1, 0], [0, 2]),
    "Partially similar": ([2, 3, 1], [1, 4, 2]),
}

for label, (first, second) in test_cases.items():
    score = calculate_cosine_similarity(first, second)
    print(f"{label:25s} -> {score:.4f}")
```
